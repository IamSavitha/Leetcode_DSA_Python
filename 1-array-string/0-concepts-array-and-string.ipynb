## Part 1 · Big-O in plain English

When an interviewer asks *"what's the time complexity?"*, they're asking:

> **If the input gets bigger, how much more work does my code do?**

We write this with **Big-O**, where `n` is the size of the input (for example, how many items are in the list).

| Big-O | Name | Plain English | Example | n = 1,000 → roughly |
|---|---|---|---|---|
| **O(1)** | constant | Same work no matter how big the input | Look up `nums[5]` | 1 step |
| **O(log n)** | logarithmic | Each step throws away half of what's left | Binary search in a sorted list | ~10 steps |
| **O(n)** | linear | Touch each item about once | Sum all numbers | 1,000 steps |
| **O(n log n)** | "n log n" | The cost of a good general sort | `sorted(nums)` | ~10,000 steps |
| **O(n²)** | quadratic | For each item, look at every item again | Compare all pairs | 1,000,000 steps |
| **O(2ⁿ)** | exponential | Try every combination | Every subset | more than the atoms in the universe |

**The everyday version:**
- **O(1)**: grabbing the book at shelf position 5. You walk straight to it.
- **O(n)**: reading every book title on the shelf once.
- **O(n²)**: for every book, comparing it with every other book.
- **O(log n)**: finding a word in a dictionary by opening it in the middle and discarding the half it can't be in.

**Space complexity** asks the same question about **memory**: *"how much extra memory do I need as the input grows?"* "O(1) extra space" means a fixed handful of variables, no matter the input size.

**Rule of thumb for interviews (about 10⁸ simple steps per second is a rough budget):**

| If n is up to… | You can usually afford… |
|---|---|
| ~10 to 20 | O(2ⁿ), trying every combination |
| ~1,000 | O(n²) |
| ~100,000 to 1,000,000 | O(n log n) or O(n) |
| bigger | O(n), O(log n) or O(1) |

This is why every problem notebook first states the **constraints**: they tell you how fast your solution needs to be.

````python
# Feel the difference: count the steps each approach takes for growing n
for n in [10, 100, 1000]:
    linear = n                     # O(n)
    quadratic = n * n              # O(n^2)
    import math
    nlogn = round(n * math.log2(n))
    print(f"n={n:>5}:  O(n)={linear:>7}   O(n log n)={nlogn:>7}   O(n^2)={quadratic:>9}")
````

## Part 2 · What is an array?

An **array** is a **row of numbered boxes** sitting side by side in memory. Each box holds one value.

```
index:    0     1     2     3     4
        ┌─────┬─────┬─────┬─────┬─────┐
nums =  │  7  │  1  │  5  │  3  │  6  │
        └─────┴─────┴─────┴─────┴─────┘
```

Three facts explain almost every cost you'll see:

1. **Boxes are numbered from 0.** The first box is index `0` and the last is index `len − 1`. ("Off-by-one" bugs come from forgetting this.)
2. **Jumping to any box is instant, O(1).** Because the boxes are side by side and equal-sized, the computer can calculate where box `i` is (`start + i × box_size`) and go straight there. This is called **random access**.
3. **Inserting or deleting in the middle is slow, O(n).** There are no gaps, so to insert at index 1 every box after it must shift one place right. To delete, they shift left to close the gap. It's like inserting someone into the middle of a queue: everyone behind them takes a step back.

**What about adding at the end?** Python's `list` is a **dynamic array**. It secretly reserves some extra empty boxes at the end. Adding to the end usually just fills the next spare box, which is O(1). When the spare boxes run out, Python allocates a bigger block (about 1.125× the size plus a little) and copies everything over. That copy is O(n), but it happens so rarely that **on average** `append` is still O(1). We call this **amortized O(1)**: an occasional expensive step, averaged over many cheap ones.

> In Python, the word **"list"** means this dynamic array. In interviews, "array" and "list" usually mean the same thing when you're writing Python.

## Part 3 · Working with Python lists

### 3.1 Creating a list

````python
empty = []                          # an empty list
nums = [7, 1, 5, 3, 6]              # list with values
zeros = [0] * 5                     # five zeros: [0, 0, 0, 0, 0]
counting = list(range(5))           # [0, 1, 2, 3, 4]
squares = [x * x for x in range(5)] # "list comprehension": build from a rule
evens = [x for x in range(10) if x % 2 == 0]   # comprehension with a filter
mixed = [1, "two", 3.0, [4]]        # lists can hold anything (but keep them uniform in practice)

print(empty, nums, zeros, counting, squares, evens, mixed, sep="\n")
````

**List comprehension** in plain English: `[x * x for x in range(5)]` reads as *"make a list of `x * x` for every `x` from 0 to 4"*. It's a compact `for` loop that builds a list.

### 3.2 Reading: indexing and length

- `nums[i]` gets the value in box `i`. **O(1).**
- **Negative indexes count from the end**: `nums[-1]` is the last item and `nums[-2]` the second-to-last. Very handy.
- `len(nums)` is the number of items. **O(1)**, because Python stores the length rather than counting.
- Asking for a box that doesn't exist raises an `IndexError`.

````python
nums = [7, 1, 5, 3, 6]
print("first:", nums[0])
print("third:", nums[2])
print("last:", nums[-1])
print("second-to-last:", nums[-2])
print("length:", len(nums))
print("last index is len - 1 =", len(nums) - 1)

try:
    nums[5]                     # there is no box 5 (valid indexes are 0..4)
except IndexError as e:
    print("IndexError:", e)
````

### 3.3 Updating a value

`nums[i] = x` overwrites box `i`. **O(1).** The list's length doesn't change.

**Swapping** two boxes is a one-liner in Python, and you'll use it constantly (reversing, rotating, two pointers):

````python
nums = [7, 1, 5, 3, 6]
nums[1] = 100                   # overwrite box 1
print(nums)

nums[0], nums[4] = nums[4], nums[0]   # swap first and last
print("after swap:", nums)
````

### 3.4 Adding items

| Operation | What it does | Cost | Why |
|---|---|---|---|
| `nums.append(x)` | add `x` at the **end** | **O(1)** amortized | Fills the next spare box |
| `nums.extend(other)` | add every item of `other` at the end | O(k), k = len(other) | One append per item |
| `nums.insert(i, x)` | put `x` at index `i` | **O(n)** | Everything from `i` onwards shifts right |
| `nums + other` | **new** list joining both | O(n + k) | Copies both lists |

👉 **Rule:** add at the **end** whenever you can. If you keep inserting at the front, you're probably using the wrong tool (see `collections.deque` in 3.11).

````python
nums = [1, 2, 3]
nums.append(4)              # [1, 2, 3, 4]
print("append:", nums)
nums.extend([5, 6])         # [1, 2, 3, 4, 5, 6]
print("extend:", nums)
nums.insert(0, 0)           # every item shifts right by one
print("insert at front:", nums)
combined = [1, 2] + [3, 4]  # a brand-new list
print("plus:", combined)
````

### 3.5 Removing items

| Operation | What it does | Cost | Why |
|---|---|---|---|
| `nums.pop()` | remove and return the **last** item | **O(1)** | Nothing needs to shift |
| `nums.pop(i)` | remove and return the item at index `i` | **O(n)** | Items after `i` shift left |
| `nums.remove(x)` | remove the **first** box equal to `x` | **O(n)** | Must search, then shift |
| `del nums[i]` | delete box `i` | O(n) | Shift left |
| `del nums[a:b]` | delete a range | O(n) | Shift left |
| `nums.clear()` | remove everything | O(n) | |

👉 **Rule:** removing from the **end** is cheap; removing from the **front or middle** is expensive. Questions 2–4 are all about avoiding that cost by **overwriting** instead of deleting.

````python
nums = [10, 20, 30, 20, 40]
last = nums.pop()           # 40 removed from the end
print("pop():", last, nums)
first = nums.pop(0)         # 10 removed from the front (others shift)
print("pop(0):", first, nums)
nums.remove(20)             # removes only the FIRST 20
print("remove(20):", nums)
del nums[0]
print("del nums[0]:", nums)

try:
    nums.remove(999)        # not present
except ValueError as e:
    print("ValueError:", e)
````

### 3.6 Searching

| Operation | Cost | Notes |
|---|---|---|
| `x in nums` | **O(n)** | Checks boxes one by one until found |
| `nums.index(x)` | O(n) | Position of the first match; `ValueError` if missing |
| `nums.count(x)` | O(n) | Always scans the whole list |
| `min(nums)`, `max(nums)`, `sum(nums)` | O(n) | Each looks at every item |
| `bisect.bisect_left(sorted_nums, x)` | **O(log n)** | Binary search, **only on sorted lists** |

**Why is `in` slow on a list but fast on a set?** A list has no idea where `x` might be, so it must look in every box. A **set** (or dict) uses hashing to jump straight to where `x` would be, which is O(1) on average. If you keep asking "have I seen this?", reach for a `set`. You'll study these properly in the Hashmap section.

````python
import bisect

nums = [7, 1, 5, 3, 6]
print("5 in nums?", 5 in nums)
print("index of 3:", nums.index(3))
print("count of 7:", nums.count(7))
print("min/max/sum:", min(nums), max(nums), sum(nums))

sorted_nums = sorted(nums)                  # [1, 3, 5, 6, 7]
pos = bisect.bisect_left(sorted_nums, 5)    # where 5 is (or would be inserted)
print("sorted:", sorted_nums, "-> 5 is at index", pos)

seen = set(nums)                            # O(n) to build, then O(1) lookups
print("6 in set?", 6 in seen)
````

### 3.7 Looping, the many ways

| You need… | Write | Example use |
|---|---|---|
| just the values | `for x in nums:` | sum, max |
| the index **and** the value | `for i, x in enumerate(nums):` | "where did I see it?" |
| only indexes (to compare neighbours) | `for i in range(1, len(nums)):` | `nums[i] - nums[i-1]` |
| backwards | `for i in range(len(nums) - 1, -1, -1):` or `for x in reversed(nums):` | fill from the back (Q1) |
| two lists side by side | `for a, b in zip(xs, ys):` | pairwise compare |
| an index that you move yourself | `while i < len(nums):` | two pointers |

**`range(start, stop, step)`** counts from `start` up to **but not including** `stop`. So `range(len(nums) - 1, -1, -1)` goes `4, 3, 2, 1, 0`.

````python
nums = [7, 1, 5, 3, 6]

for i, x in enumerate(nums):
    print(f"index {i} holds {x}")

print("neighbour differences:", [nums[i] - nums[i - 1] for i in range(1, len(nums))])
print("backwards indexes:", list(range(len(nums) - 1, -1, -1)))
print("reversed values:", list(reversed(nums)))
print("zipped:", list(zip(["a", "b", "c"], [1, 2, 3])))
````

### 3.8 Slicing: taking a piece of a list

`nums[start:stop:step]` makes a **new list** from `start` up to (not including) `stop`. Any part can be left out.

| Slice | Meaning |
|---|---|
| `nums[1:4]` | indexes 1, 2, 3 |
| `nums[:3]` | the first 3 items |
| `nums[2:]` | from index 2 to the end |
| `nums[-2:]` | the last 2 items |
| `nums[::2]` | every second item |
| `nums[::-1]` | reversed copy |
| `nums[:]` | a full (shallow) copy |

⚠️ **A slice is a copy**, so it costs **O(k)** time and memory for k items. Slicing inside a loop can quietly turn O(n) code into O(n²).

**Slice assignment** replaces a section in place: `nums[1:3] = [9, 9]`. Question 6 (Rotate Array) uses `nums[:] = ...` to overwrite the whole list *in place*. `nums = ...` would only point the local name at a new list.

````python
nums = [0, 1, 2, 3, 4, 5]
print(nums[1:4], nums[:3], nums[2:], nums[-2:], nums[::2], nums[::-1])

nums[1:3] = [9, 9]          # replace a section
print("slice-assigned:", nums)

nums[:] = [5, 4, 3]         # replace the contents in place (same list object)
print("whole replaced:", nums)
````

### 3.9 Sorting

| | `nums.sort()` | `sorted(nums)` |
|---|---|---|
| Changes the original? | **Yes**, in place | **No**, returns a new list |
| Returns | `None` (a classic bug: `x = nums.sort()` makes `x` None) | the sorted list |
| Cost | O(n log n) | O(n log n) + O(n) memory for the copy |

Both accept `reverse=True` and `key=...` (a function that says *what* to sort by). Python's sort is **stable**: items that compare equal keep their original order.

````python
nums = [7, 1, 5, 3, 6]
print("sorted copy:", sorted(nums), "original:", nums)
nums.sort(reverse=True)
print("sorted in place, descending:", nums)

words = ["banana", "kiwi", "apple", "fig"]
print("by length:", sorted(words, key=len))                 # stable: ties keep order
print("by last letter:", sorted(words, key=lambda w: w[-1]))

result = nums.sort()
print("nums.sort() returns:", result)                       # None, the classic trap
````

### 3.10 Copying: alias vs. copy (a very common source of bugs)

`b = a` does **not** copy the list. It gives the **same list** a second name (an **alias**). Changing `b` changes `a`.

| Code | What you get |
|---|---|
| `b = a` | same list, two names |
| `b = a[:]`, `a.copy()`, `list(a)` | a **shallow copy**: a new outer list, but inner objects are shared |
| `copy.deepcopy(a)` | a fully independent copy, inner lists included |

````python
import copy

a = [1, 2, 3]
b = a            # alias
b.append(4)
print("alias changed a too:", a)

c = a[:]         # copy
c.append(5)
print("copy left a alone:", a, "| c =", c)

grid = [[1, 2], [3, 4]]
shallow = grid[:]
deep = copy.deepcopy(grid)
shallow[0][0] = 99
print("shallow copy shares inner lists:", grid)
print("deep copy is independent:", deep)
````

### 3.11 2D lists (grids) and `deque`

**Grids** are lists of lists: `grid[row][col]`.

⚠️ **The famous trap:** `[[0] * 3] * 3` creates **one** inner list and repeats a reference to it 3 times. Changing one row changes all three. Always build grids with a comprehension: `[[0] * cols for _ in range(rows)]`.

**`collections.deque`** ("double-ended queue", pronounced "deck") is the right tool when you need to add or remove at **both** ends. `appendleft` and `popleft` are O(1) there, while on a list they're O(n).

````python
from collections import deque

bad = [[0] * 3] * 3
bad[0][0] = 1
print("trap: ", bad)                      # all rows changed!

good = [[0] * 3 for _ in range(3)]
good[0][0] = 1
print("right:", good)

q = deque([2, 3])
q.appendleft(1)        # O(1)
q.append(4)            # O(1)
print("deque:", q, "| popleft ->", q.popleft(), "| now", q)
````

## Part 4 · What is a string?

A **string** is a **row of characters**, just like an array of letters:

```
index:   0   1   2   3   4
s =     'h' 'e' 'l' 'l' 'o'
```

So reading `s[i]`, `len(s)`, and slicing all work exactly like lists.

**The one big difference: Python strings are immutable** (they can't be changed after creation).
- `s[0] = "H"` is an **error**.
- Every "change" (`s.upper()`, `s.replace(...)`, `s + "!"`) builds a **new** string, which costs O(n).

**Why it matters:** building a string piece by piece with `+=` inside a loop can copy the growing string over and over, which is O(n²) in the worst case. The fix is to **collect the pieces in a list and `"".join(...)` them once at the end**, which is O(n). (CPython sometimes optimises `+=`, but you can't rely on that, and interviewers know it.)

**Characters are numbers underneath.** Each character has a code: `ord("a") == 97`, and `chr(97) == "a"`. That's how you turn letters into array indexes: `ord(ch) - ord("a")` gives 0 for `a` up to 25 for `z`.

````python
s = "hello"
print(s[0], s[-1], len(s), s[1:4], s[::-1])

try:
    s[0] = "H"
except TypeError as e:
    print("TypeError:", e)

t = "H" + s[1:]               # build a NEW string instead
print("new string:", t, "| original unchanged:", s)

print("ord('a') =", ord("a"), "| chr(98) =", chr(98), "| index of 'c' =", ord("c") - ord("a"))
````

### 5.3 Searching inside a string

| Method | Returns | Cost |
|---|---|---|
| `sub in s` | True/False | O(n·m) worst case (usually fast in practice) |
| `s.find(sub)` | first index, or **−1** if missing | same |
| `s.index(sub)` | first index, or **raises `ValueError`** | same |
| `s.count(sub)` | number of non-overlapping matches | O(n) |
| `s.startswith(p)` / `s.endswith(p)` | True/False | O(len(p)) |

Question 23 asks you to implement `find` yourself (brute force and KMP).

````python
s = "sadbutsad"
print("'but' in s:", "but" in s)
print("find('sad'):", s.find("sad"), "| find('xyz'):", s.find("xyz"))
print("count('sad'):", s.count("sad"))
print("startswith('sad'):", s.startswith("sad"), "| endswith('bad'):", s.endswith("bad"))
````

### 5.4 Split and join: turning text into lists and back

- `s.split()` with **no argument** splits on **any run of whitespace** and drops empty pieces. This is what you want for "words".
- `s.split(" ")` splits on **each single space**, so double spaces produce empty strings `""`. This is a common bug.
- `"sep".join(list_of_strings)` glues pieces together with `sep` in between. It's O(total length) and the **right way to build strings**.
- `list(s)` turns a string into a list of characters (mutable!), and `"".join(chars)` turns it back.

````python
s = "  the sky   is blue "
print(s.split())             # ['the', 'sky', 'is', 'blue']
print(s.split(" "))          # empty strings appear!
print(" ".join(["a", "b", "c"]), "|", "-".join("abc"))

chars = list("hello")        # mutable copy
chars[0] = "j"
print("".join(chars))        # 'jello'
````

### 5.5 Building strings efficiently

The bad way (can be O(n²)) vs. the good way (O(n)):

````python
words = ["build", "strings", "with", "join"]

# ❌ may copy the growing result on every +=
out = ""
for w in words:
    out += w + " "
print(repr(out.strip()))

# ✅ collect pieces, join once at the end
parts = []
for w in words:
    parts.append(w)
print(repr(" ".join(parts)))

# also handy: repetition and f-strings
print("ab" * 3, "|", f"{len(words)} words, first is {words[0]!r}")
````

### 5.6 Replacing, reversing, comparing, counting letters

- `s.replace(old, new)` → new string with every `old` replaced. O(n).
- `s[::-1]` → reversed copy. O(n).
- Strings compare **alphabetically** (character code by character code): `"apple" < "banana"` is True, but `"Z" < "a"` is also True, because uppercase codes are smaller.
- **Counting letters**: a fixed-size list of 26 counts (using `ord`), or `collections.Counter` (a dict of counts, which you'll meet in the Hashmap section).

````python
from collections import Counter

print("hello world".replace("o", "0"))
print("racecar"[::-1] == "racecar")
print("apple" < "banana", "Z" < "a")

word = "banana"
counts = [0] * 26
for ch in word:
    counts[ord(ch) - ord("a")] += 1
print("array counts:", {chr(i + ord('a')): c for i, c in enumerate(counts) if c})
print("Counter:", Counter(word))
````

## Part 6 · Cost cheat sheet

**Python `list`** (n = length)

| Operation | Cost | Remember it as |
|---|---|---|
| `a[i]`, `a[i] = x`, `len(a)` | O(1) | jump straight to a box |
| `a.append(x)`, `a.pop()` | O(1) amortized | work at the end is cheap |
| `a.insert(i, x)`, `a.pop(i)`, `del a[i]`, `a.remove(x)` | O(n) | the middle/front forces shifting |
| `x in a`, `a.index(x)`, `a.count(x)`, `min/max/sum` | O(n) | must look at every box |
| `a[i:j]` (slice) | O(j − i) | it's a copy |
| `a.sort()`, `sorted(a)` | O(n log n) | general sorting |
| `a.reverse()`, `a[::-1]` | O(n) | touch every box |
| `bisect` on a sorted list | O(log n) | halve each step |

**Python `str`** (n = length)

| Operation | Cost | Remember it as |
|---|---|---|
| `s[i]`, `len(s)` | O(1) | same as a list |
| slice, `+`, `upper`, `lower`, `strip`, `replace`, `[::-1]` | O(n) | every change makes a new string |
| `"".join(parts)` | O(total length) | **the** way to build strings |
| `s.split()` | O(n) | |
| `sub in s`, `s.find(sub)` | O(n·m) worst | usually fast in practice |
| `+=` in a loop | up to O(n²) total | avoid it; use a list + join |

**Other structures you'll meet in these problems**

| Structure | Strength | Typical use here |
|---|---|---|
| `set` | O(1) average "have I seen x?" | remove duplicates, membership |
| `dict` | O(1) average key → value lookup | value → index (Q12), counting |
| `collections.deque` | O(1) at both ends | queues, sliding windows |
| `collections.Counter` | counting in one line | Majority Element (Q5), anagrams |

## Part 7 · Common traps (read these twice)

1. **Off-by-one.** The last index is `len(a) − 1`. `range(n)` stops *before* `n`.
2. **Changing a list while looping over it.** Removing items inside `for x in a:` skips elements. Loop over a copy, or use a read/write pointer (Q2).
3. **Alias instead of copy.** `b = a` shares the list. Use `a[:]` or `a.copy()`.
4. **`[[0] * m] * n` grid trap.** All rows are the same object. Use a comprehension.
5. **`x = a.sort()`** gives `None`. Use `sorted(a)` if you need the result.
6. **`s.split(" ")` vs `s.split()`.** Only the second collapses multiple spaces.
7. **String `+=` in a loop.** Collect pieces in a list, then `"".join`.
8. **`nums = something` inside a function** doesn't change the caller's list. Use `nums[:] = something` to overwrite in place (Q6).

````python
# Trap 2 demonstrated: removing while iterating skips items
a = [2, 2, 3, 2]
for x in a:
    if x == 2:
        a.remove(x)
print("buggy remove-while-looping:", a)        # a 2 survives!

a = [2, 2, 3, 2]
a = [x for x in a if x != 2]                     # safe: build a new list
print("safe:", a)

# Trap 8 demonstrated
def reassign(nums):
    nums = [0, 0, 0]          # only changes the local name

def overwrite(nums):
    nums[:] = [0, 0, 0]       # changes the caller's list

x = [1, 2, 3]; reassign(x);  print("after reassign:", x)
x = [1, 2, 3]; overwrite(x); print("after overwrite:", x)
````

## Part 8 · Patterns you'll use in Questions 1–24

These are the reusable ideas. Each problem notebook will name its pattern, and you can jump back here to revise it.

### Pattern A · Read/write pointers (in-place filtering)
**Idea:** one index **reads** every item and another **writes** the keepers to the front. Nothing is deleted, so nothing shifts.
**Used in:** Q2 Remove Element, Q3 & Q4 Remove Duplicates.

````python
def keep_odds(nums):
    write = 0
    for read in range(len(nums)):
        if nums[read] % 2 == 1:
            nums[write] = nums[read]
            write += 1
    return write                    # how many were kept

a = [1, 2, 3, 4, 5]
k = keep_odds(a)
print(k, a[:k])
````

### Pattern B · Two pointers from opposite ends
**Idea:** start one index at each end and move them toward each other. Great for reversing, palindromes, and pair problems on sorted data.
**Used in:** Q1 Merge Sorted Array (from the back), Q6 Rotate Array (reverse), Q16 Trapping Rain Water, Q21 Reverse Words (in place).

````python
def reverse_in_place(a, lo, hi):
    while lo < hi:
        a[lo], a[hi] = a[hi], a[lo]
        lo += 1
        hi -= 1

a = [1, 2, 3, 4, 5]
reverse_in_place(a, 0, len(a) - 1)
print(a)
````

### Pattern C · Running "best so far" (greedy one pass)
**Idea:** walk once, keeping a small summary of everything seen so far (minimum, maximum, farthest reach…), and update the answer as you go.
**Used in:** Q7 & Q8 Stock problems (running minimum price), Q9 & Q10 Jump Game (farthest reachable), Q14 Gas Station (running tank).

````python
prices = [7, 1, 5, 3, 6, 4]
min_price, best = float("inf"), 0
for p in prices:
    min_price = min(min_price, p)        # cheapest day so far
    best = max(best, p - min_price)      # best profit if we sell today
print("best single trade profit:", best)
````

### Pattern D · Prefix and suffix computations
**Idea:** precompute "everything to my left" and "everything to my right" for each index in two passes. Then each answer is O(1).
**Used in:** Q13 Product Except Self, Q15 Candy, Q16 Trapping Rain Water.

````python
nums = [1, 2, 3, 4]
prefix_sum = [0] * (len(nums) + 1)          # prefix_sum[i] = sum of nums[:i]
for i, x in enumerate(nums):
    prefix_sum[i + 1] = prefix_sum[i] + x
print("prefix sums:", prefix_sum)
print("sum of nums[1:3] in O(1):", prefix_sum[3] - prefix_sum[1])
````

### Pattern E · Counting with an array or dict
**Idea:** when values are small integers (or letters), a list of counts indexed by value replaces sorting or searching.
**Used in:** Q5 Majority Element (count / Boyer–Moore vote), Q11 H-Index (bucket counts).

````python
citations = [3, 0, 6, 1, 5, 3]
count = [0] * (max(citations) + 1)
for r in citations:
    count[r] += 1
print("count per value:", count)
````

### Pattern F · Reverse tricks
**Idea:** reversing the whole thing and then parts of it moves blocks around in place.
**Used in:** Q6 Rotate Array (reverse all, then each part), Q21 Reverse Words (reverse all, then each word).

````python
# rotate right by k using three reversals
def rotate(nums, k):
    n = len(nums)
    k %= n
    def rev(lo, hi):
        while lo < hi:
            nums[lo], nums[hi] = nums[hi], nums[lo]
            lo, hi = lo + 1, hi - 1
    rev(0, n - 1)      # [7,6,5,4,3,2,1]
    rev(0, k - 1)      # [5,6,7,4,3,2,1]
    rev(k, n - 1)      # [5,6,7,1,2,3,4]

a = [1, 2, 3, 4, 5, 6, 7]
rotate(a, 3)
print(a)
````

### Pattern G · Hash map + array together
**Idea:** an array gives fast access **by position**; a dict gives fast access **by value**. Keep both in sync to get both.
**Used in:** Q12 Insert Delete GetRandom O(1).

**Idea:** some problems have no trick, just rules to follow exactly. Write small helper steps and test the edge cases.
